Person sits at top left corner of an m * n of grid .It can only move right or down .How many unique paths exist to reach bottom right corner ?

House_robber_problem --> CLASSIC DP

```python
def rob_dp(nums):
    n = len(nums)
    if n == 0:
        return 0
    if n == 1:
        return nums[0]
    dp = [0] * n
    dp[0] = nums[0]
    dp[1] = max(nums[0], nums[1])

    for i in range(2, n):
        dp[i] = max(dp[i-1], nums[i] + dp[i-2])
    return dp[-1]
print(rob_dp([2,8,9,12]))
```

SPACE OPTIMIZED HOUSE ROBBER PROBLEM


```python
def rob_optimized(nums):
    prev2, prev1 = 0, 0  # dp[i-2], dp[i-1]

    for num in nums:
        current = max(prev1, num + prev2)
        prev2 = prev1
        prev1 = current

    return prev1
```

MIN COST CLIMBING STAIRS


```python
def min_cost_climbing_stairs(cost):
    n = len(cost)
    if n == 0:
        return 0
    if n == 1:
        return 0  # top is one step away, free to "reach" trivially

    dp = [0] * n
    dp[0] = cost[0]
    dp[1] = cost[1]

    for i in range(2, n):
        dp[i] = cost[i] + min(dp[i-1], dp[i-2])

    return min(dp[-1], dp[-2])
```

SPACE OPTIMIZED 1D DP

```python
def unique_paths_1d(m, n):
   dp = [1] * n

   for i in range(1, m):
       for j in range(1, n):
           dp[j] += dp[j-1]
   return dp[-1]
print(unique_paths_1d(3, 7))
print(unique_paths_1d(3, 2))
```

LONGEST STRICTLY INCREASING SUBSEQUENCE FROM GIVEN INTERGER DP


```python
def length_of_lis(nums):
    if not nums:
        return 0

    n = len(nums)
    dp = [1] * n   # every element is a subsequence of length 1 by itself

    for i in range(1, n):
        for j in range(i):
            if nums[j] < nums[i]:
                dp[i] = max(dp[i], dp[j] + 1)

    return max(dp)
```

OPTIMIZED O(NLOGN) BINARY SEARCH


```python
from bisect import bisect_left

def length_of_lis_fast(nums):
    tails = []

    for num in nums:
        pos = bisect_left(tails, num)
        if pos == len(tails):
            tails.append(num)      # extends the longest subsequence
        else:
            tails[pos] = num       # replaces to keep tail as small as possible

    return len(tails)
```

RETURN THE ACTUAL COIN USED NOT JUST THE COIN


```python
def coin_change_with_coins(coins, amount):
    dp = [float('inf')] * (amount + 1)
    dp[0] = 0
    used_coin = [-1] * (amount + 1)   # tracks which coin was used at each amount

    for i in range(1, amount + 1):
        for c in coins:
            if c <= i and dp[i - c] + 1 < dp[i]:
                dp[i] = dp[i - c] + 1
                used_coin[i] = c

    if dp[amount] == float('inf'):
        return -1, []

    # backtrack to reconstruct the coins
    result = []
    curr = amount
    while curr > 0:
        result.append(used_coin[curr])
        curr -= used_coin[curr]

    return dp[amount], result
```
